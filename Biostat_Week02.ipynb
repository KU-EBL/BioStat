```python
import statistics
import numpy as np
```

```python
data1 = [
    7.2, 7.6, 8.5, 8.35, 8.7, 9, 9, 9.3, 9.4,
    10.2, 10.9, 11.3, 12.1, 12.8
]


mode = statistics.mode(data1)
median = statistics.median(data1)
mean = statistics.mean(data1)
maximum = max(data1)
minimum = min(data1)

print("최빈값:", mode)
print("중앙값:", median)
print("평균:", round(mean, 2))
print("최댓값:", maximum)
print("최솟값:", minimum)
```

```python
data2 = [
    21.2, 21.6, 21.7, 23.1, 23.7,
    24.2, 24.2, 25.5, 26.6, 26.7,
    26.8, 28.9, 29.0, 29.7, 29.7,
    30.2, 32.5, 33.7, 33.7
]

Q1, Q2, Q3 = np.percentile(data2, [25, 50, 75])

print("최솟값:", min(data2))
print("제1사분위수(Q1):", Q1)
print("중앙값(M):", Q2)
print("제3사분위수(Q3):", Q3)
print("최댓값:", max(data2))
```

```python
data3 = [1792, 1666, 1362, 1614, 1460, 1867, 1439]

sample_variance = statistics.variance(data3)
sample_std = statistics.stdev(data3)

print("표본분산:", round(sample_variance, 2))
print("표본표준편차:", round(sample_std, 2))
```
